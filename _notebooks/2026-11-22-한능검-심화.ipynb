## 삼국 시대

### 고구려

`-` 진대법, 태학과 경당(인재 양성, 소수림왕)

`-` 불교: 쌍영총벽화, 승려 순도, 소수림왕(고국원왕의 아들)

`-` 장군총(돌무지 무덤)

### 백제

`-` 마지막 수도 부여, 삼국시대에 부여, 좌평 얘기나오면 백제임

`-` 6좌평(달솔 등의 관등)

`-` 근초고왕: 고국원왕 kill (그래서 보복으로 장수왕한테 개로왕 전사), 한강차지, 서기 편찬(한강 차지~)

`-` 무령왕: 22담로, 백가의 난 진압, 중국 남조

`-` 성왕: 사비 천도, 국호 남부여, 진흥왕 뒤통수로 나제동맹 결렬(관산성(충북) 전투에서 전사), 달기를 보내 고구려의 도살성을 함락

`-` 무왕: 미륵사

`-` 송산리 고분(무령왕릉), 능산리 고분(사비, 부여)

`-` 고구려에 태학과 경당이 있고 신라에 국학이 있다면 백제엔 오경박사(유교)가 있따

### 신라

`-` 지증왕(22대): 국호 신라, 동시전 설치, 우경 장려, 우산국 복속

`-` 법흥왕: 병부와 상대등 설치(관청, 중앙집권체제), 율법과 불교(이차돈 순교), 건원 연호

`-` 진흥왕: 화랑도(게이화장) 개편, 거칠부로 하여금 국사편찬, 나제 동맹 배신 깜, 4개의 순수비

`-` 문무왕: 백제 부흥 진압(복신, 도침), 고구려 부흥 지원(안승을 보덕국 왕으로), 외사정을 파견(지방관 감시), 상수리제도(기인제도로 발전)

`-` 신문왕(통일신라): 녹읍 폐지(대신 관료전 지급, 진골 귀족 조질려고), 김흠돌의 난 진압(통일직후), 국학(박사와 조교를 두어 유학 교육), 9주5소경, 9서당

`-` 강수(문장가): 무열왕~문무왕, 6두품(관등제한 벽느낌. max불가능), 청방인문표(외교 문서)

`-` 설총: 이두(우리말 표기법), 아빠가 원효대사 해골물, 신문왕한테 화왕계 들려주~

`-` 원광(승려): 걸사표, 세속5계

`-` 원효대사(마음먹기에 달림): 불교 대중화, 화쟁사상, 일심사상, 금강산매경론 저술

`-` 의상(승려): 당으로 건너가 불법, 화엄종, 관음신앙, 화엄일승법계도

`-` 혜초(승려): 왕오천축국전(인도여행, 구법순례기)

`-` 자장(승려): 황룡사9층목탑

`-` 골품제

`-` 화랑도(얼굴에 분칠, 청소년 민간 단체)

### 신라의 삼국 통일 과정

`-` 살수대첩(을지문덕) -> 천리장성 축조 -> (배경) 백제 의자왕 신라의 대양성 공격 -> 막리지의 난(연개소문 정변, 영류왕 시해) -> 안시성전투(연개소문, 보장왕) -> 김춘추(무열왕) 개빡침 -> 나당동맹(당나라는 고구려한테 빡침) -> 백제 공격 ㄱㄱ 황산벌 전투에서 승리 -> 흑치상지의 백제 부흥 운동 시발점(임존성에서 군사를 일으킴), 복신과 도침 지도자(**부여**풍 왕으로) -> 나당연합의 백강 전투 승리 (백제 부흥 운동의 끝자락) -> 고구려 공격 (연개소문 없으니까 썰어버림) -> 당나라의 안동도호부 -> 고구려 부흥 운동 지원 (신라가 지원해줌, 안승을 보덕국 왕으로) -> 당나라가 뒤통수 깜, 매기전투(매소성, 기벌포)에서 신라 승리 -> 통일~

### 통일 신라

`-` 수도에 서시와 남시 설치

### 후삼국

`-` 궁예(후고구려, 신라 왕족 출신): 광평성(정치기구), 국호를 마진과 태봉으로, 철원으로 천도

`-` 견훤(후백제, 신라말기 군인 출신): 완산주에 나라 세움, 후당 오월에 사신 파견(겸교태보의 직을 받음), 태조 왕건에게 항복gg

`-` 이색: 성리학 연구, 목은집, 공민왕 때 성균관 대사성이 되어 정몽주 학관으로 천거(select)

`-` 호헌(전땅크 무한 대통령, 간선제) 철폐와 독재 타도, 삼청교육대, 국보위, 야간 통행 가능하고 두발의 자유화~, 이산가족상봉, 3저 호황과 최저임금, 부산 경찰서 성고문, 보도지침으로 언론 통제, 서울 아시아 게임, 경부고속도로

### 노태우 (1988 ~ 1993)

`-` 서울 올림픽(88올림픽), 7.7선언, 남북 고위급 선언, 남북한 un동시 가입, 남북 기본 합의서, 민주화시위(강경대 학생 희생)/민주 자유당(3당 합당), 한반도 비핵화 공동 선언

### 김영삼 (1993 ~ 1998)

`-` 금융실명제, 거창사건명예회복특별법, IMF, OECD, 하나회 척결, 문민정부(드디어 대통령이 군인이 아님), 역사 바로 세우기로 조선 총독부(무단 통치) 철거/국민학교를 초등학교로, 전국민노총;

### 김대중 (1998 ~ 2003)

`-` IMF 해결, FIFA 한일 월드컵, 국가 인권 위원회와 국민 기초 생활 보장법, 최초의 남북정상회담(1차, 6.15), 개성공업지구 조성 합의, 금강산 관강 스타투, 노사정 위원회(대통령 직속 자문 기구)

### 노무현 (2003 ~ 2008)

`-` 10.4 남북정상선언(2차), 국가 원수로서는 최초로 군사 분계선을 도보 통과, APEC 정상 회의, 개성공단 착공, 금강산 육로, 호주제 폐지, 친일 반민족 행위 진상 규명
